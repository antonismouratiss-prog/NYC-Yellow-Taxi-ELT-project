# Bronze Layer — Raw Ingestion
## NYC Yellow Taxi — January 2024

This notebook downloads the raw TLC Yellow Taxi Parquet file for January 2024
and saves it as-is into the Bronze Delta table. No transformations are applied.

**Source:** NYC TLC Trip Record Data  
**Target:** `nyc_taxi.bronze.yellow_trips_raw`

```python
import subprocess

url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-01.parquet"
volume_path = "/Volumes/nyc_taxi/bronze/raw_files/yellow_tripdata_2024-01.parquet"

subprocess.run(["wget", "-q", url, "-O", volume_path], check=True)
print("Download complete.")
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

df_raw = spark.read.parquet(volume_path)
```

```python
df_raw.count()
```

```python
(df_raw.write.format("delta").saveAsTable("nyc_taxi.bronze.yellow_trips_raw"))
```
